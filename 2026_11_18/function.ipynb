### 引數位置(position Argument)
最常見的呼叫方式，就是直接把值放進括號裡。Python 會依照呼叫時排列的先後順序，從左到右一對一對應到函式定義的參數位置：

```python
def describe_pet(animal, name):
    print(f"我養了一隻 {animal}，它的名字是 {name}。")

describe_pet("狗", "小白")  
describe_pet("小白", "狗")
#describe_pet("狗")
#describe_pet("狗", "小白", 1)
```

### 引數名稱(keyword Argument)
當函式的參數比較多、不容易記住順序時，可以在呼叫時直接寫出 參數名稱=值，這種方式稱為引數名稱呼叫（或關鍵字引數）：

```python
def describe_pet(animal, name):
    print(f"我養了一隻 {animal}，它的名字是 {name}。")

describe_pet(animal="狗", name="小白")
describe_pet(name="小白", animal="狗")
```

### 重要規則:混用時 位置引數 一定要在前面

```python
describe_pet(animal="狗", name="小白")
```

### 指配預設值(Default Argument) 
定義函式時，如果有幫參數設定預設值，呼叫時就能只改自己想調整的項目：

```python
def order_drink(drink, ice="正常冰", sugar="半糖"):
    print(f"你點了一杯 {drink}，冰塊量為 {ice}，甜度為 {sugar}。")

order_drink("咖啡")
order_drink("咖啡", "少冰")
order_drink("咖啡", sugar="無糖")
order_drink("咖啡", sugar="無糖", ice="少冰")
```

## 特殊符號 / 與 * —— 限制引數的呼叫方式
### /(僅限位置引數 Positional-Only)
寫在 `/` 左邊的所有參數，呼叫時只能依位置傳入，不能寫出引數名稱：

## 2. *（僅限關鍵字引數 Keyword-Only）
寫在單獨一個 * 右邊的所有參數，呼叫時必須寫出引數名稱，不能只靠位置傳入：

```python
def cale_bmi(w, h , /):
    bmi = w / (h ** 2)
    print(f"你的BMI為 {bmi:.2f}")   

cale_bmi(70, 1.75)
```

```python
def send_email(to, *, subject, urgent=False):
    print(f"寄信給 {to}，主旨為 {subject}，是否緊急：{urgent}")

send_email("hero63@gmail.com", subject="會議通知", urgent=True)
send_email("hero63@gmail.com", urgent=True, subject="會議通知")
```

## *args 與 **kwargs —— 定義函式時接收任意數量的引數（打包 Packing）
有時候設計函式時，並不知道使用者會傳入幾個引數（例如計算任意個數字的總和，或接收額外的設定欄位）。這時候可以在定義函式時，在參數前面加上 * 或 **，把多出來的引數「打包」起來。

### *args（一個星號）：把多餘的「位置引數」打包成 tuple

### **kwargs （兩個星號）：把多餘的「引數名稱」打包成 dict

```python
def sum_scores(*scores):
    print("收到", scores, type(scores))
    return sum(scores)

print(sum_scores(90, 80, 70))
print(sum_scores(90, 80))
print(sum_scores())
```

```python
def build_profile(name, age, **info):
    print(f"姓名：{name}｜年齡：{age}")
    print("其他資訊：", info, type(info))

build_profile("小明", 18, city="台北", hobby="打籃球", play="電腦遊戲")
```
